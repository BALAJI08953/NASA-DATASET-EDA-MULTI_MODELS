```python
#1 MISSING VALUES

#or the list comprention
dataset_na=[dataset for dataset in df_train.columns if df_train[dataset].isnull().sum()>1]

for dataset in dataset_na:
    print(dataset,np.round(df_train[dataset].isnull().mean(),4),'%missing values')
```

```python
#2 ALL THE NUMERIC VARIABLES
df_train.info()
```

```python
# SINCE THEY ARE MANY MISSIN VALUES WE NEED TO FIND THE REALTIONSHIP BETWEEN MISSING VALUES AND SALES PRICE
for dataset in dataset_na:
    data = df_train.copy()

    data[dataset] = np.where(data[dataset].isnull(), 1, 0)

    print(dataset)
    print(data[dataset].value_counts())

    print(data.groupby(dataset)['SalePrice'].median())

    data.groupby(dataset)['SalePrice'].median().plot.bar()
    plt.title(dataset)
    plt.show()
  
```

```python
#numerical features
numerical_features=[game for game in df_train.columns if df_train[game].dtype !='O']
print('NUMBER OF NUMERICAL VARIBLES:',len(numerical_features))
```

```python
df_train[numerical_features].head()
```

```python
year_feature = [
    col for col in numerical_features
    if 'yr' in col.lower() or 'year' in col.lower()
]

print(year_feature)
```

```python
for col in year_feature:
    print(col,df_train[col].unique())
```

```python
#lets analyze the temporal date time and variables
df_train.groupby('YrSold')['SalePrice'].median().plot()
plt.xlabel('Yr_Sold')
plt.ylabel('Meadian House Price')
plt.title('HOUSE PRICE AND YEARSOLD')
```

```python
for game in year_feature:
    if game!='YrSold':
        data=df_train.copy()
        data[game]=data['YrSold']-data[game]
        #we will capture the diifernce between year variable and year the year_build
        plt.scatter(data[game],data['SalePrice'])
        plt.xlabel(game)
        plt.ylabel('SalePrice')
        plt.show()
```

```python
#numerical variables are usally in 2 types 
## continous variable and discrete variable
disc_features=[disc for disc in numerical_features if len(df_train[disc].unique())<25 and disc not in year_feature+['Id'] ]
print('discrete variable count:{}'.format(len(disc_features)))
```

```python
disc_features
```

```python
#
df_train[disc_features].head()
```

```python
fig, axes = plt.subplots(len(disc_features), 1, figsize=(12, 5*len(disc_features)))

i = 0

for disc in disc_features:
    data = df_train.copy()

    data.groupby(disc)['SalePrice'].median().plot.bar(ax=axes[i])

    axes[i].set_xlabel(disc)
    axes[i].set_ylabel('SalePrice')
    axes[i].set_title(disc)

    i = i + 1

plt.tight_layout()
plt.show()
```

```python
#continous variable
continous_features=[features for features in numerical_features if features not in disc_features +year_feature+['Id']]
print(len(continous_features))
```

```python
for feature in continous_features:
    data=df_train.copy()
    data[feature].hist(bins=25)
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.ylabel(feature)
    plt.title(feature)
    plt.show()
```
